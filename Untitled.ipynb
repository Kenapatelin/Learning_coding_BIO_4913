## *Slicing is splitting a string into characters. You can index into a string just like you can index into a list (or nested list). We can slice one character, or many characters)*

```python
print(FIRST)
print('The last letter of my name is:', FIRST[3])
```

```python
# Slicing many characters from a string at once
FULLNAME="Kena Patel"
print(FULLNAME)
NEW_FIRST=FULLNAME[0:4] #characters 0 through 3, not including 5
NEW_LAST=FULLNAME[5:10] #characters 0 through 4, not including 11
print('First name:',NEW_FIRST)
print('Last name:',NEW_LAST)
```

```python
#create a nested slice
ELEMENTS=[['fluorine', 'F'],
          ['chlorine', 'Cl'],
          ['bromine', 'Br'],
          ['hydrogen', 'H'],
          ['helium', 'He']]
```

```python
print(ELEMENTS)
```
